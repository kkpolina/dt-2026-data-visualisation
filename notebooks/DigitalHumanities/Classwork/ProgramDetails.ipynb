```python
import pandas as pd
df = pd.read_csv("Competencies1.csv")
df.head() # Show the first few rows
df.describe() # Show some basic statistics
```

```python
import matplotlib.pyplot as plt
avg_hours = df.groupby("Student")["Hours_Studied"].mean()
#colors= plt.vm.plasma(range(len(avg_hours)))
avg_hours.plot(kind="bar",x="Student",y="Hours_Studies", color="#E4A0F7")
plt.show()

```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6, color="#A7C7E7")
plt.title("Grade vs Hours Studied")
plt.ylabel("Grade")
plt.xlabel("Hours Studied")

```

```python
plt.Figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Distribution of grades and study hours")
plt.show
```
