```python
import pandas as pd
df = pd.read_csv("Movies.csv")
df.head()
df.describe()
```

```python
import matplotlib.pyplot as plt
country_count = df.groupby("Type")["Country"].count()
country_count.plot(kind="bar", color="#AFE1AF")
plt.xlabel("Type")
plt.ylabel("Number of Countries")
plt.title("Number of Countries by Type")
plt.show()
```

Reflection: The chart shows that Movies and TV Shows have the same number of country entries. This means that both types are equally represented in my dataset.

```python
plt.scatter(df["Release Year"], df["Country"].str.split(",").str.len(), s=df["Release Year"]*0.1, alpha=0.6, color="#90EE90")
plt.title("Number of Countries vs Release Year")
plt.ylabel("Number of Countries")
plt.xlabel("Release Year")
```

Reflection: The chart shows the relationship between release year and the number of countries. Most of the titles were released in 2021 and have one country listed. Some titles have more than one country, but there does not seem to be a clear relationship between release year and the number of countries.

```python
plt.Figure(figsize=(8,6))
df.boxplot(column="Release Year", by="Type", color="#009E60")
plt.title("Distribution of release years by type")
plt.suptitle("")
plt.xlabel("Type")
plt.ylabel("Release Year")
plt.show()
```

Reflection: The chart shows the distribution of release years for Movies and TV Shows. Most of the titles in both types were released in 2021. Movies have a wider range of release years, while TV Shows are more concentrated around recent years.

```python
rating_count = df["Rating"].value_counts()
rating_count.plot(kind="pie", labels=rating_count.index, legend=False, autopct="%1.1f%%")
plt.title("Distribution of Ratings")
plt.ylabel("")
plt.show()
```

Reflection: The chart shows the distribution of ratings in the dataset. TV-MA is the most common rating, while some ratings appear less often. Overall, the dataset contains a mix of different age ratings.
